```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 72
from IPython.display import Markdown
import numpy as np
import os

MODEL_PATH = '/content/EP_models/'
os.environ['HF_HOME'] = MODEL_PATH
os.environ['HF_DATASETS_OFFLINE'] = '0'  # 0 = allow download first time

import torch
import transformers
from transformers import pipeline

Device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print(f'PyTorch version= {torch.__version__}')
print(f'transformers version= {transformers.__version__}')
print(f'CUDA available= {torch.cuda.is_available()}')
```

```python
from huggingface_hub import login
login()
```

```python
Model = pipeline("text-generation", model="microsoft/phi-4", max_length=4096, dtype=torch.bfloat16, device=Device)
```

```python
def query(_model, _prompt):
    messages = [{'role':'user', 'content':_prompt}]
    response = _model(messages)
    return response[0]['generated_text'][1]['content']
```

```python
Markdown(query(Model, 'Who are you?'))
```

```python
Markdown(query(Model, 'The sky is ...'))
```

```python
Markdown(query(Model, 'Be creative from now on. The sky is...'))
```

```python
# text summarization
prompt = """In the vast universe of Star Trek, the franchise explores themes that transcend time, blending futuristic technology
with profound philosophical dilemmas. One of the most iconic elements is the Prime Directive, a rule forbidding Starfleet from
interfering with the natural development of alien civilizations. This principle often places characters in morally complex situations,
such as Captain Jean-Luc Picard’s struggle to balance his moral compass with the cold impartiality of Starfleet regulations. Meanwhile,
the series has introduced groundbreaking technology, from replicators that eliminate scarcity to warp drives enabling interstellar travel,
reflecting humanity's aspirations and fears about technological advancement. Through its diverse crews and stories, Star Trek
champions a hopeful vision of the future, where cooperation and understanding pave the way for progress across the galaxy.

Explain the above in one short sentence:"""

Markdown(query(Model, prompt))
```

```python
# question answering
prompt = """Answer the question based on the context below. Keep the answer short and concise.
Respond "Unsure about answer" if not sure about the answer.

Context: Teplizumab traces its roots to a New Jersey drug company called Ortho Pharmaceutical. There, scientists generated an early
version of the antibody, dubbed OKT3. Originally sourced from mice, the molecule was able to bind to the surface of T cells and limit
their cell-killing potential. In 1986, it was approved to help prevent organ rejection after kidney transplants, making it the first
therapeutic antibody allowed for human use.

Question: What was OKT3 originally sourced from?

Answer:"""

Markdown(query(Model, prompt))
```

```python
# text classification
prompt = """Classify the text into neutral, negative, or positive.

Text: I think the food was so so.

Sentiment:"""

Markdown(query(Model, prompt))
```

```python
# role playing
prompt = """The following is a conversation with an AI research assistant. The assistant tone is technical and scientific.

Human: Hello, who are you?
AI: Greeting! I am an AI research assistant. How can I help you today?
Human: Can you tell me about the creation of blackholes?
AI:"""

Markdown(query(Model, prompt))
```

```python
# code generation
prompt = """How can I write a Python program from scratch that generates a random Star Trek-style
spaceship name using a list of adjectives and nouns combined with simple string manipulation and randomness?
"""

Markdown(query(Model, prompt))
```

```python
# reasoning
prompt = """The odd numbers in this group add up to an even number: 15, 32, 5, 13, 82, 7, 1.

Solve by breaking the problem into steps. First, identify the odd numbers, add them, and indicate whether the result is odd or even."""

Markdown(query(Model, prompt))
```

```python
prompt= """Explain the philosophical implications of Gödel's incompleteness theorems on the feasibility of a unified theory of everything in physics,
referencing connections to quantum mechanics, the limits of human cognition, and the nature of mathematical truth—all within 100 words."""

Markdown(query(Model, prompt))
```
